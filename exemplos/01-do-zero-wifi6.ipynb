# Do zero: a troca de roteadores por Wi-Fi 6 valeu a pena?

Este caderno mostra o fluxo principal do ADEX Copiloto: você chega **só com a pergunta**. O exemplo usa dados
sintéticos locais (DuckDB), então roda sem a OCI. Precisa do hub (`qwen-latest`) e da chave conferida por impressão:

```bash
adex-segredo definir hub.chave
adex-configuracoes iniciar-local
```

Em cada ⛔ quem decide é você.

````python
from adex_copiloto import AdexCopiloto

adex = AdexCopiloto.from_env()
ex = adex.exemplo("wifi6")          # 41 cidades, ondas de troca, códigos de chamado (sintético)
estudo = adex.create_session(objective=ex.objetivo, data_sources=ex.fonte(adex.config))
estudo.frame()                       # enquadra: nenhum dado lido, até 3 perguntas
````

Responda do seu jeito, numa rodada só. O que você disser literalmente vira **você disse ✓** na carta.

````python
carta = estudo.frame(answers="""1. Decidir se a onda 4 sai em 2027 e se vale para todas as cidades.
2. Chamados técnicos de Wi-Fi por mil clientes.
3. Cidades que ainda não trocaram servem de comparação.""")
carta
````

````python
carta.approve()                      # ⛔ carta do estudo
````

````python
estudo.discover(hint="acho que tem algo no esquema OPS_RES")   # só metadados
````

````python
estudo.confirm_mapping()
````

````python
estudo.decisions()                    # decisões em aberto (ex.: códigos em dúvida)
````

Se houver decisão antes do universo, responda com `estudo.decide('D-01', 'a', reason='...')`.

````python
estudo.run("universo")              # painel agregado na fonte, pelo porteiro
````

````python
estudo.passport().approve()           # ⛔ Passaporte de Amostragem
````

````python
plano = estudo.plan()                 # jeitos de responder + método por blocos + pré-mortem
plano
````

````python
plano.approve()                       # ⛔ plano
````

````python
estudo.run()                          # etapas do plano (blocos registrados)
````

````python
estudo.review()                       # crítico (Python + Qwen) e narrativa com números do Python
````

````python
estudo.accept(reason="efeito consistente; placebo e sensibilidade conferidos")   # ⛔ aceite
````

````python
h = estudo.export_handoff(incluir_dados="agregados")
h
````

````python
h.approve()                           # ⛔ publica o pacote que roda sem o copiloto
````
